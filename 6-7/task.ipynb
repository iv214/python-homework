```python
import requests as rq
import pandas as pd
import os
import re
from dotenv import load_dotenv

load_dotenv()
```

#### Задание 1.
Получение данных из публичного API
1. Выберите публичный API. Например, [JSONPlaceholder](https://jsonplaceholder.typicode.com/).
2. Напишите скрипт, который:
    * отправляет GET-запрос к /posts,
    * извлекает и выводит заголовки и тела первых 5 постов.

```python
URL = "https://jsonplaceholder.typicode.com/posts"

response = rq.get(url=URL)
print(f"Status code: {response.status_code} {response.reason}")
```

```python
result = [
    {"Title": post["title"], "Body": post["body"]}
    for post in response.json()[:5]
]
pd.DataFrame(result).style.hide(axis="index")
```

#### Задание 2. Работа с параметрами запроса
1. Используйте API OpenWeather для получения данных о погоде.
2. Напишите программу, которая:
* принимает название города от пользователя,
* отправляет GET-запрос к API и выводит текущую температуру и описание погоды.

```python
__API_key = os.getenv("OPENWEATHERMAP_API_KEY")
OWM_URL = "https://api.openweathermap.org/data/2.5/weather"
```

```python
city_name = re.sub(r"[/\\?%*:|\"<>;=]", "", input().strip())
params = {
    "q": city_name,
    "appid": __API_key,
    "units": "metric",
}
try:
    owm_response = rq.get(url=OWM_URL, params=params)
except Exception as e:
    print(type(e).__name__)
del params
print(f"Status code: {owm_response.status_code} {owm_response.reason}")
```

```python
owm_data = owm_response.json()
owm_result = [
    {
        "City": city_name,
        "Temperature, °C": owm_data["main"]["temp"],
        "Description": owm_data["weather"][0]["description"],
    }
]
pd.DataFrame(owm_result).style.hide(axis="index").format(precision=2)
```
